```python
torch.manual_seed(123)
import torch
import torch.nn as nn
torch.set_printoptions(sci_mode=False)
```

```python
batch_example = torch.randn(2,5)
layer = nn.Sequential(nn.Linear(5, 6), nn.ReLU())
out = layer(batch_example)
print(out)
```

```python
mean = out.mean(dim=1, keepdim=True)
var = out.var(dim=1, keepdim=True)
print("Mean:", mean)
print("Variance:", var)
```

- The 1st row in the mean tensor above contains the mean value for the first input row, and the 2nd output row contains the mean of 2nd input row. 

- Using Keepdim = True, in operations like mean or variance calculation ensures that the output tensor retain the same number of dimensions as the input term, even though the operation reduces the tensor along the dimension specified via dim.

- For instance, without keepdim=TRUE, the returned mean tensor would be a 2-dimensional vector [0,1324, 0,2170], instead of 2x1-dimensional matrix [[0.1324], [0.2170]]

- For a 2-D tensor (like a matrix), using dim=-1 for operations such as mean or variance calculation is the same as using dim=1

- This is because -1 refers to the tensor's last dimension, which corresponds to the columns in a 2D tensor.

- Later, when adding layer normalisation to the GPT model, which produces 3D tensors with shape [batch_size, num_tokens, embedding_size] we can still use dim=-1 for normalization across the last dimension, avoiding a change from dim=1 to dim=2.

#### Apply layer norm

- Next, let us apply layer normalization to the layer outputs we obtained earlier.

- The operation consist of subtracting the mean and divide by sqr. root of variance (std deviation)

```python
out_norm = (out - mean) / torch.sqrt(var)
mean = out_norm.mean(dim=1, keepdim=True)
var = out_norm.var(dim=1, keepdim=True)
print("Normalized layer outputs :", out_norm)

print("Mean: ", mean)
print("Variance: ", var)
```

### Let's encapsulate and build the LayerNorm class

```python
class LayerNorm(nn.Module):
    def __init__(self, emb_dim):
        super().__init__()
        self.eps = 1e-5
        self.scale = nn.Parameter(torch.ones(emb_dim))
        self.shift = nn.Parameter(torch.zeros(emb_dim))

    def forward(self, x):
        mean = x.mean(dim=-1, keepdim=True) # calculate row-wise mean
        var = x.var(dim=-1, keepdim=True, unbiased=False) # calculate row-wise variance
        x_norm = (x - mean) / torch.sqrt(var + self.eps) # eps is added to avoid division by zero
        return self.scale * x_norm + self.shift
```

- This specific implementation of the layer Normalization operates on the last dimension of the input tensor x, which represent the embedding dimension (emb_dim).

- The variable eps is a small constant (epsilon) added to the variance to prevent division by ZERO during normalization.

- The SCALE and SHIFT are 2 trainable parameters (of the same dimesion as the input) that the LLM automatically adjusts during training if it is determined that doing so would improve the model's performance on its training task.

- This allows the model to learn appropriate scaling and shifting that best suit the data it is processing.
